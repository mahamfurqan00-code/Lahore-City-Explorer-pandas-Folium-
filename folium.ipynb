```python
import folium

# Example number of cases
cases = 12450

# Create map
m = folium.Map(
    location=[24.8607, 67.0011],   # Karachi
    zoom_start=11
)

# Add circle marker
folium.CircleMarker(
    location=[24.8607, 67.0011],
    radius=min(cases / 500, 30),
    color="crimson",
    fill=True,
    fill_color="crimson",
    fill_opacity=0.6,
    popup=f"Karachi: {cases} cases"
).add_to(m)

m
```

```python
import folium

# Create map
m = folium.Map(
    location=[31.5204, 74.3587],   # Lahore
    zoom_start=12,
    tiles="CartoDB Positron"
)

# Create HTML popup
html = """
<h3>Lahore</h3>
<p>Population: 13 million</p>
<p>Cases: 12,450</p>
"""

# Add marker with HTML popup
folium.Marker(
    location=[31.5204, 74.3587],
    popup=folium.Popup(html, max_width=250),
    tooltip="Click for Lahore details"
).add_to(m)

# Display map
m
```

```python
import folium

# Example cases value
cases = 12450

# Function for marker color
def severity_color(cases):
    if cases > 10000:
        return "red"
    elif cases > 1000:
        return "orange"
    return "green"

# Create map
m = folium.Map(
    location=[31.5204, 74.3587],  # Lahore
    zoom_start=12,
    tiles="CartoDB Positron"
)

# Add marker
folium.Marker(
    location=[31.5204, 74.3587],
    popup=f"Lahore: {cases} cases",
    tooltip="Severity Level",
    icon=folium.Icon(
        color=severity_color(cases),
        icon="exclamation-triangle",
        prefix="fa"
    )
).add_to(m)

# Display map
m
```
